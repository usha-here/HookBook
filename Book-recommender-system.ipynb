```python
def recommend(book_name):
    # Case-insensitive partial match for book titles
    matching_books = final_ratings[final_ratings['Book-Title'].str.contains(book_name, case=False, na=False)]

    if matching_books.empty:
        return f"No books found containing the word: '{book_name}'"

    # Group by Book-Title, Book-Author, and Image, then calculate average rating
    grouped = matching_books.groupby(['Book-Title', 'Book-Author', 'Image-URL-M']).agg({'Book-Rating': 'mean'}).reset_index()

    # Sort by average rating in descending order
    sorted_books = grouped.sort_values(by='Book-Rating', ascending=False)

    # Prepare the final result
    data = []
    for _, row in sorted_books.iterrows():
        item = [row['Book-Title'], row['Book-Author'], row['Image-URL-M'], round(row['Book-Rating'], 2)]
        data.append(item)

    return data
```

```python
nltk.download('punkt')
stemmer = PorterStemmer()

def stem_text(text):
    """Stem all words in the text."""
    words = re.findall(r'\w+', text.lower())
    return ' '.join(stemmer.stem(w) for w in words)
```

```python
def recommend(book_name):
    # Stem the search term
    search_term_stemmed = stem_text(book_name)

    # Stem book titles for matching
    final_ratings['stemmed_title'] = final_ratings['Book-Title'].apply(stem_text)

    # Filter where stemmed title contains the stemmed search term
    matching_books = final_ratings[
        final_ratings['stemmed_title'].str.contains(search_term_stemmed, na=False)
    ]

    if matching_books.empty:
        return f"No books found containing or similar to: '{book_name}'"

    # Group by original title, author, and image; calculate average rating
    grouped = matching_books.groupby(
        ['Book-Title', 'Book-Author', 'Image-URL-M']
    ).agg({'Book-Rating': 'mean'}).reset_index()

    # Sort by average rating (highest first)
    sorted_books = grouped.sort_values(by='Book-Rating', ascending=False)

    # Prepare result
    data = []
    for _, row in sorted_books.iterrows():
        data.append([
            row['Book-Title'],
            row['Book-Author'],
            row['Image-URL-M'],
            round(row['Book-Rating'], 2)
        ])

    return data
```

```python
recommend('computer')
```
